```python
# ==========================================
# PHASE 2: DAILY HISTORY FETCH + STORAGE
#           (FINNHUB, DAILY, REFRESHABLE)
# ==========================================

import os

DATA_PATH = "/content/drive/MyDrive/SMA_Swing_Project/daily_history_live.parquet"
# DATA_PATH_1 = "/content/drive/MyDrive/SMA_Swing_Project/daily_history_250.parquet"
# DATA_PATH_2 = "/content/drive/MyDrive/SMA_Swing_Project/daily_history_300.parquet"
# DATA_PATH_3 = "/content/drive/MyDrive/SMA_Swing_Project/daily_history_350.parquet"
# DATA_PATH_4 = "/content/drive/MyDrive/SMA_Swing_Project/daily_history_500.parquet"
# DATA_PATH_5 = "/content/drive/MyDrive/SMA_Swing_Project/daily_history_600.parquet"

# -----------------------------------------------------------
# 1️⃣ Build unified daily dataset for many symbols
# -----------------------------------------------------------
def fetch_daily_history_finnhub(
    symbols,
    lookback_days=365*3,
    end_date=None,
    data_path=DATA_PATH,
    verbose_failures=True,
):
    """
    Fetch ~lookback_days of daily candles for each symbol via Finnhub
    and store them in a single Parquet file.

    - Uses your existing fetch_finnhub_daily(symbol, lookback_days)
    - Schema per row:
        Date (tz-aware, TIMEZONE)
        Ticker
        Open, High, Low, Close, Adj Close, Volume
        Ret_1, LogRet_1, DayOfWeek, DayIndex
    - If data_path already exists:
        - Drop existing rows for refreshed symbols
        - Append new data, sort, and overwrite file
    """
    all_data = []
    failures = []

    for s in tqdm(symbols, desc="Fetching daily data from Finnhub"):
        df, status = fetch_finnhub_daily(s, lookback_days=lookback_days, end_date=end_date)
        if df is None or df.empty:
            failures.append((s, status))
            continue

        # Ensure expected columns, add Adj Close
        if "Adj Close" not in df.columns:
            df["Adj Close"] = df["Close"]

        # Light daily returns
        df["Ret_1"] = df["Close"].pct_change()
        df["LogRet_1"] = np.log(df["Close"]).diff()

        # Time-based features
        df["DayOfWeek"] = df["Date"].dt.dayofweek  # 0=Mon ... 6=Sun

        # Monotonic trading-day index per ticker (for future horizon calc if needed)
        df["DayIndex"] = np.arange(len(df))

        # Column order for cleanliness
        cols = [
            "Date", "Ticker",
            "Open", "High", "Low", "Close", "Adj Close", "Volume",
            "Ret_1", "LogRet_1", "DayOfWeek", "DayIndex",
        ]
        df = df[cols]

        all_data.append(df)

    if not all_data:
        raise ValueError("No daily historical data fetched from Finnhub.")

    new_data = pd.concat(all_data, ignore_index=True)

    # Merge with existing file if present
    try:
        existing = pd.read_parquet(data_path)
        # Drop rows for tickers we are refreshing
        existing = existing[~existing["Ticker"].isin(symbols)]
        combined = pd.concat([existing, new_data], ignore_index=True)
    except FileNotFoundError:
        combined = new_data

    combined = combined.sort_values(["Ticker", "Date"]).reset_index(drop=True)

    # Make sure folder exists
    os.makedirs(os.path.dirname(data_path), exist_ok=True)
    combined.to_parquet(data_path, index=False)

    if verbose_failures and failures:
        print("\n⚠️ Finnhub daily fetch failures (first 20):")
        print(failures[:20])
        if len(failures) > 20:
            print(f"... and {len(failures) - 20} more")

    print(f"\n✅ Saved daily dataset to {data_path}")
    print(f"   Rows: {len(combined):,} | Cols: {len(combined.columns)}")
    print(f"   Tickers stored: {combined['Ticker'].nunique()}")

    return combined


# -----------------------------------------------------------
# 2️⃣ Simple loader
# -----------------------------------------------------------
def load_daily_history(path=DATA_PATH):
    df = pd.read_parquet(path)
    # Ensure Date is datetime with timezone
    df["Date"] = pd.to_datetime(df["Date"])
    if df["Date"].dt.tz is None:
        df["Date"] = df["Date"].dt.tz_localize(TIMEZONE)
    return df.sort_values(["Ticker", "Date"]).reset_index(drop=True)


# -----------------------------------------------------------
# 3️⃣ PHASE 2 sanity check
# -----------------------------------------------------------
def phase2_sanity_check():
    """
    Quick check using a tiny subset of SYMBOLS:
      - Run fetch_daily_history_finnhub on 2–3 tickers
      - Inspect shapes, columns, and per-ticker date ranges
    """
    test_syms = ["NVDA", "MSFT", "AAPL"]
    print("🚀 Phase 2 sanity check: building daily dataset\n")

    hist = fetch_daily_history_finnhub(
        test_syms,
        lookback_days=365*1,
        #end_date="2025-01-01",
        data_path=DATA_PATH,
        verbose_failures=True,
    )

    print("\n✅ Preview (head):")
    print(hist.head(10))

    print("\n✅ Preview (tail):")
    print(hist.tail(10))

    print("\n✅ Columns:")
    print(list(hist.columns))

    print("\n✅ Per-ticker ranges:")
    for s in test_syms:
        g = hist[hist["Ticker"] == s]
        if g.empty:
            print(f"{s}: no rows stored.")
        else:
            print(f"{s}: {len(g):,} rows | {g['Date'].min()} → {g['Date'].max()}")


# Uncomment to run Phase 2 sanity check
#phase2_sanity_check()

# When you're happy, run full universe:


# Run Phase 2 fetch once to build the base daily dataset
hist_daily = fetch_daily_history_finnhub(
    SYMBOLS,
    lookback_days=300,      # or 3, or whatever you chose earlier
    #end_date="2025-01-01",
    data_path=DATA_PATH,
    verbose_failures=True,
)
print("\n✅ Preview (tail):")
print(hist_daily.head(10))
print(hist_daily.tail(10))





# When you're happy, run full universe:
#full_hist = fetch_daily_history_finnhub(SYMBOLS, lookback_days=365*5)
```

```python
# ==========================================
# PHASE 3: DAILY SWING FEATURES + FLEXIBLE HORIZON LABEL
# ==========================================
import os

DATA_PATH = "/content/drive/MyDrive/SMA_Swing_Project/daily_history_live.parquet"
DAILY_FEATURE_PATH = "/content/drive/MyDrive/SMA_Swing_Project/daily_features_live.parquet"

def add_daily_technical_indicators(
    daily_df,
    sma_windows_short=(5, 10, 20, 30),
    sma_windows_long=(50,60, 100),
    ema_windows=(5,10,20),
    vol_windows=(10, 20, 50),
    rsi_period=14,
    bb_window=20,
    bb_k=1.0,
):
    """
    Per-ticker daily indicators.
    Input:  columns: Date, Ticker, Open/High/Low/Close/Volume
    Output: same rows + indicator columns
    """
    df = daily_df.sort_values(["Ticker", "Date"]).copy()
    out_frames = []

    for ticker, g in df.groupby("Ticker"):
        g = g.sort_values("Date").copy()

        # daily returns
        g["ret_1d"] = g["Close"].pct_change()
        g["log_ret_1d"] = np.log(g["Close"]).diff()

        # SMAs
        for w in sma_windows_short + sma_windows_long:
            g[f"sma_{w}"] = g["Close"].rolling(w).mean()

        for w in ema_windows:
            g[f"ema_{w}"] = g["Close"].ewm(span=w, adjust=False).mean()

        # vol of returns
        for w in vol_windows:
            g[f"vol_{w}"] = g["ret_1d"].rolling(w).std()

        # Bollinger
        g[f"bb_mid_{bb_window}"] = g["Close"].rolling(bb_window).mean()
        g[f"bb_std_{bb_window}"] = g["Close"].rolling(bb_window).std()
        g[f"bb_up_{bb_window}"]  = g[f"bb_mid_{bb_window}"] + bb_k * g[f"bb_std_{bb_window}"]
        g[f"bb_dn_{bb_window}"]  = g[f"bb_mid_{bb_window}"] - bb_k * g[f"bb_std_{bb_window}"]

        # RSI
        delta = g["Close"].diff()
        gain = delta.clip(lower=0).rolling(rsi_period).mean()
        loss = (-delta.clip(upper=0)).rolling(rsi_period).mean()
        rs = gain / loss
        g[f"rsi_{rsi_period}"] = 100 - (100 / (1 + rs))

        out_frames.append(g)

    full = pd.concat(out_frames, ignore_index=True)
    return full


def build_and_save_daily_features(
    input_path,
    output_path=DAILY_FEATURE_PATH,
    sma_windows_short=(5, 10, 20),
    sma_windows_long=(50, 100),
    ema_windows=(5,10,20),
    vol_windows=(10, 20, 50),
    rsi_period=14,
    bb_window=20,
    bb_k=2.0,
    max_horizon=10,
    flex_start=5,
    flex_end=10,
):
    """
    Phase 3:
      1) Load daily history from input_path
      2) Add indicators
      3) Add fut_ret_1 ... fut_ret_max_horizon
      4) Build flexible horizon label:
           fut_ret_5_10_max     = best future return in [5..10] days
           fut_ret_5_10_best_h  = horizon (in days) that achieved that best return
         Rows with no future data (all NaN) stay NaN and will be dropped later.
      5) Save to output_path
    """
    if not os.path.exists(input_path):
        raise FileNotFoundError(f"{input_path} does not exist. Run Phase 2 first.")

    daily_df = pd.read_parquet(input_path)
    daily_df["Date"] = pd.to_datetime(daily_df["Date"])

    print(f"✅ Loaded daily history from {input_path}")
    print(f"   Rows: {len(daily_df):,} | Cols: {daily_df.shape[1]}")

    # 1) Add indicators
    feats = add_daily_technical_indicators(
        daily_df,
        sma_windows_short=sma_windows_short,
        sma_windows_long=sma_windows_long,
        ema_windows=ema_windows,
        vol_windows=vol_windows,
        rsi_period=rsi_period,
        bb_window=bb_window,
        bb_k=bb_k,
    )
    print(feats[["Ticker", "Date", "Close",
                 "sma_5","ema_5"]].tail(10))


    # 2) Future returns per ticker
    feats = feats.sort_values(["Ticker", "Date"]).reset_index(drop=True)
    out_frames = []

    tickers = feats["Ticker"].unique()
    for ticker in tickers:
        g = feats[feats["Ticker"] == ticker].copy()
        g = g.sort_values("Date").reset_index(drop=True)

        # fut_ret_h: Close[t+h]/Close[t] - 1
        for h in range(1, max_horizon + 1):
            g[f"fut_ret_{h}"] = g["Close"].shift(-h) / g["Close"] - 1.0

        out_frames.append(g)

    feats = pd.concat(out_frames, ignore_index=True)

    # 3) Flexible horizon label [flex_start..flex_end]
    flex_cols = [f"fut_ret_{h}" for h in range(flex_start, flex_end + 1)]
    flex_mat = feats[flex_cols].to_numpy(dtype=float)
    n = flex_mat.shape[0]
    horizons = np.arange(flex_start, flex_end + 1)

    # mask of rows that have at least one non-NaN in [flex_start..flex_end]
    valid_mask = ~np.isnan(flex_mat).all(axis=1)

    max_vals = np.full(n, np.nan, dtype=float)
    best_h_vals = np.full(n, np.nan, dtype=float)

    # Only compute nanmax/nanargmax on rows that actually have data
    if valid_mask.any():
        sub = flex_mat[valid_mask]
        max_vals[valid_mask] = np.nanmax(sub, axis=1)
        best_idx = np.nanargmax(sub, axis=1)
        best_h_vals[valid_mask] = horizons[best_idx]

    feats["fut_ret_5_10_max"] = max_vals
    feats["fut_ret_5_10_best_h"] = best_h_vals

    # Optional: drop rows where we *never* have future info at all
    # (you can also leave them; Phase 4 will drop NaNs in the label)
    # feats = feats.dropna(subset=["fut_ret_5_10_max"])

    # Save
    os.makedirs(os.path.dirname(output_path), exist_ok=True)
    feats.to_parquet(output_path, index=False)

    print(f"\n✅ Saved daily swing features to {output_path}")
    print(f"   Rows: {len(feats):,} | Cols: {feats.shape[1]}")

    return feats


# -----------------------------------------------------------
# PHASE 3 SANITY CHECK
# -----------------------------------------------------------
def phase3_sanity_check():
    print("🚀 Phase 3 sanity check: building daily swing features\n")

    feats = build_and_save_daily_features(
        input_path=DATA_PATH,
        output_path=DAILY_FEATURE_PATH,
        sma_windows_short=(5, 10, 20),
        sma_windows_long=(50, 100),
        ema_windows=(5,10,20),
        vol_windows=(10, 20, 50),
        rsi_period=14,
        bb_window=20,
        bb_k=2.0,
        max_horizon=10,
        flex_start=5,
        flex_end=10,
    )

    print("\n✅ Feature preview:")
    print(feats[["Ticker", "Date", "Close",
                 "fut_ret_5", "fut_ret_5_10_max", "fut_ret_5_10_best_h"]].tail(10))

    print("\n📊 Label stats (fut_ret_5_10_max):")
    print(feats["fut_ret_5_10_max"].describe(percentiles=[0.1, 0.25, 0.5, 0.75, 0.9]))

    print("\n📊 Best horizon distribution (5 to 10 days):")
    print(feats["fut_ret_5_10_best_h"].value_counts(dropna=True).sort_index())


# Uncomment to run sanity check once
phase3_sanity_check()
```

```python
# ==========================================
# PHASE 4: XGBOOST TRAINING FOR SWING MODEL
#          (FLEXIBLE 5–10 DAY HORIZON)
# ==========================================
import os
import json
import pickle
import numpy as np
import pandas as pd
from xgboost import XGBRegressor

# Paths (match Phase 3)
DAILY_FEATURE_PATH = "/content/drive/MyDrive/SMA_Swing_Project/daily_features_live.parquet"
SWING_MODEL_PATH   = "/content/drive/MyDrive/SMA_Swing_Project/xgb_swing_model_live.pkl"
SWING_META_PATH    = "/content/drive/MyDrive/SMA_Swing_Project/xgb_swing_model_meta_live.json"

# Label settings
FLEX_LABEL_COL   = "fut_ret_5_10_max"      # from Phase 3
FLEX_H_RANGE     = (5, 10)                 # for meta info only
VAL_FRAC         = 0.2                     # last 20% of time as validation


# -----------------------------------------------------------
# 1️⃣ Load feature dataset (Phase 3 output)
# -----------------------------------------------------------
def load_daily_features(path=DAILY_FEATURE_PATH):
    if not os.path.exists(path):
        raise FileNotFoundError(f"{path} not found. Run Phase 3 first.")

    df = pd.read_parquet(path)
    df["Date"] = pd.to_datetime(df["Date"])
    df = df.sort_values(["Ticker", "Date"]).reset_index(drop=True)

    print(f"✅ Loaded daily features from {path}")
    print(f"   Rows: {len(df):,} | Cols: {df.shape[1]}")
    return df


# -----------------------------------------------------------
# 2️⃣ Build train/valid matrices and train XGBoost
# -----------------------------------------------------------
def train_swing_xgb(
    feats_df,
    label_col=FLEX_LABEL_COL,
    val_frac=VAL_FRAC,
):
    """
    Train an XGBoost regression model to predict `label_col`
    (here: fut_ret_5_10_max = best return in 5–10 days ahead).

    Returns:
      model, feature_cols, (train_df, val_df)
    """
    if label_col not in feats_df.columns:
        raise ValueError(f"{label_col} not found in features_df. Did you run Phase 3 correctly?")

    df = feats_df.copy()
    df = df.sort_values(["Date", "Ticker"]).reset_index(drop=True)

    # Drop rows with no label (e.g., last days of each ticker)
    df = df.replace([np.inf, -np.inf], np.nan)
    df = df.dropna(subset=[label_col])

    # Optional: drop any remaining NaNs anywhere
    df = df.dropna().reset_index(drop=True)

    print("\n🧹 After cleaning:")
    print(f"   Rows: {len(df):,} | Cols: {df.shape[1]}")

    # Time-based split (global, across all tickers)
    n = len(df)
    split_idx = int(n * (1 - val_frac))
    train_df = df.iloc[:split_idx].copy()
    val_df   = df.iloc[split_idx:].copy()

    print("\n🧪 Time split:")
    print(f"   Split index: {split_idx} / {n}")
    print(f"   Train period: {train_df['Date'].min()} → {train_df['Date'].max()}")
    print(f"   Valid period: {val_df['Date'].min()} → {val_df['Date'].max()}")
    print(f"   Train rows: {len(train_df):,}")
    print(f"   Valid rows: {len(val_df):,}")

    # Feature selection:
    # drop Date, Ticker, all fut_ret_* columns, and helper horizon column
    drop_cols = ["Date", "Ticker"]
    drop_cols += [c for c in df.columns if c.startswith("fut_ret_")]
    drop_cols += ["fut_ret_5_10_best_h"]  # horizon index (if present)

    drop_cols = list(set(drop_cols))  # unique

    feature_cols = [c for c in df.columns if c not in drop_cols]
    feature_cols = [c for c in feature_cols if c != label_col]

    print("\n🔧 Feature matrix:")
    print(f"   Using {len(feature_cols)} feature columns.")
    print("   Example features:", feature_cols[:10])

    X_train = train_df[feature_cols].values
    y_train = train_df[label_col].values
    X_val   = val_df[feature_cols].values
    y_val   = val_df[label_col].values

    print(f"   X_train: {X_train.shape}, X_val: {X_val.shape}\n")

    # XGBoost model (regression on future return)
    model = XGBRegressor(
        n_estimators=400,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        tree_method="hist",
        eval_metric="rmse",
    )

    print("🚀 Training swing XGBoost model...")
    model.fit(
        X_train,
        y_train,
        eval_set=[(X_train, y_train), (X_val, y_val)],
        verbose=False,
    )

    # Predictions
    y_pred_train = model.predict(X_train)
    y_pred_val   = model.predict(X_val)

    # RMSE
    train_rmse = np.sqrt(np.mean((y_train - y_pred_train) ** 2))
    val_rmse   = np.sqrt(np.mean((y_val   - y_pred_val)   ** 2))

    # Directional accuracy on "is this a good swing?"
    # Define "positive swing" as > 0 (you can tighten later)
    train_dir = (np.sign(y_train) == np.sign(y_pred_train)).mean()
    val_dir   = (np.sign(y_val)   == np.sign(y_pred_val)).mean()

    print("\n📊 Performance:")
    print(f"   Train RMSE: {train_rmse:.4f}")
    print(f"   Valid RMSE: {val_rmse:.4f}")
    print(f"   Train dir-acc: {train_dir:.3f}")
    print(f"   Valid dir-acc: {val_dir:.3f}")

    # Sanity check: top vs bottom decile on validation
    val_df = val_df.copy()
    val_df["y_true"] = y_val
    val_df["y_pred"] = y_pred_val

    try:
        val_df["pred_decile"] = pd.qcut(val_df["y_pred"], 10, labels=False, duplicates="drop")

        top_dec = val_df[val_df["pred_decile"] == val_df["pred_decile"].max()]
        bot_dec = val_df[val_df["pred_decile"] == val_df["pred_decile"].min()]

        print("\n🔍 Sanity check (validation deciles):")
        print(f"   Top decile avg label ({label_col}): {top_dec[label_col].mean():.4%} (n={len(top_dec)})")
        print(f"   Bottom decile avg label:           {bot_dec[label_col].mean():.4%} (n={len(bot_dec)})")
    except Exception as e:
        print("\n⚠️ Could not compute deciles for sanity check:", e)

    return model, feature_cols, (train_df, val_df)


# -----------------------------------------------------------
# 3️⃣ Save model + meta
# -----------------------------------------------------------
def save_swing_model(model, feature_cols, label_col=FLEX_LABEL_COL, h_range=FLEX_H_RANGE):
    os.makedirs(os.path.dirname(SWING_MODEL_PATH), exist_ok=True)

    # Model
    with open(SWING_MODEL_PATH, "wb") as f:
        pickle.dump(model, f)

    # Meta info for Phase 5/7 live usage
    meta = {
        "feature_cols": feature_cols,
        "label_col": label_col,
        "flex_h_start": h_range[0],
        "flex_h_end": h_range[1],
    }
    with open(SWING_META_PATH, "w") as f:
        json.dump(meta, f, indent=2)

    print(f"\n💾 Saved swing model to: {SWING_MODEL_PATH}")
    print(f"💾 Saved meta to:        {SWING_META_PATH}")


# -----------------------------------------------------------
# 4️⃣ PHASE 4 DRIVER + SANITY CHECK
# -----------------------------------------------------------
def phase4_train_swing_model():
    print("🚀 Phase 4: Training daily swing XGBoost model\n")

    feats_df = load_daily_features(DAILY_FEATURE_PATH)
    model, feature_cols, (train_df, val_df) = train_swing_xgb(
        feats_df,
        label_col=FLEX_LABEL_COL,
        val_frac=VAL_FRAC,
    )
    save_swing_model(model, feature_cols, label_col=FLEX_LABEL_COL, h_range=FLEX_H_RANGE)

    print("\n✅ Phase 4 complete.")
    return model, feature_cols, (train_df, val_df)


# Run once after Phase 3 is done:
model, feature_cols, (train_df, val_df) = phase4_train_swing_model()
```

```python
import os, json, pickle
import numpy as np
import pandas as pd

# ----------------------------
# Paths (match your project)
# ----------------------------

FEATURE_PATH = "/content/drive/MyDrive/SMA_Swing_Project/daily_features_live.parquet"
MODEL_PATH   = "/content/drive/MyDrive/SMA_Swing_Project/xgb_swing_model_live.pkl"
META_PATH    = "/content/drive/MyDrive/SMA_Swing_Project/xgb_swing_model_meta_live.json"

OUT_TRADES_PATH = "/content/drive/MyDrive/SMA_Swing_Project/backtest_trades_phase5_1.csv"
OUT_EQUITY_PATH = "/content/drive/MyDrive/SMA_Swing_Project/backtest_equity_phase5_1.csv"

# ----------------------------
# Strategy parameters (tune)
# ----------------------------
BUY_THRESH     = 0.3   # enter if predicted max return >= 1.5%
STOP_LOSS      = -0.075   # hard stop at -6% from entry

TRAIL_START    = 0.1    # start trailing once trade has been +4% at any time
TRAIL_GAP      = 0.03    # allow 3% drawdown from peak once trailing is active

# Flexible horizon window used in Phase 3 labeling: [H_MIN, H_MAX]
# We do NOT force an exit at H_MAX if the trade is profitable.
H_MIN = 5
H_MAX = 7

# Fees (optional)
FEE_PER_TRADE  = 0.000   # set e.g. 0.001 for 0.1% per round trip if you want
# We'll subtract fee on exit (simple). If you want entry+exit fees, double it.

# ----------------------------
# 0) Load model + meta
# ----------------------------
def load_model_and_meta(model_path=MODEL_PATH, meta_path=META_PATH):
    with open(model_path, "rb") as f:
        model = pickle.load(f)
    with open(meta_path, "r") as f:
        meta = json.load(f)

    feature_cols = meta.get("feature_cols", None)
    if not feature_cols or not isinstance(feature_cols, list):
        raise ValueError("META missing feature_cols (list).")

    return model, meta, feature_cols


# ----------------------------
# 1) Load features
# ----------------------------
def load_daily_features(feature_path=FEATURE_PATH):
    df = pd.read_parquet(feature_path)
    # normalize Date
    if "Date" not in df.columns:
        raise ValueError("Feature dataset must include 'Date'.")
    df["Date"] = pd.to_datetime(df["Date"])
    df = df.sort_values(["Ticker", "Date"]).reset_index(drop=True)
    return df


# ----------------------------
# 2) Add predictions (pred_max)
# ----------------------------
def add_predictions(df, model, feature_cols):
    missing = [c for c in feature_cols if c not in df.columns]
    if missing:
        raise KeyError(f"Feature dataset missing required feature_cols: {missing}")

    # Ensure numeric dtype for XGBoost
    X = df[feature_cols].apply(pd.to_numeric, errors="coerce")
    # Drop rows with any NaN in features (cannot predict)
    ok = ~X.isna().any(axis=1)

    df = df.copy()
    df["pred_max"] = np.nan

    if ok.sum() == 0:
        raise ValueError("No rows have complete numeric features for prediction.")

    preds = model.predict(X.loc[ok].values)
    df.loc[ok, "pred_max"] = preds.astype(float)
    return df


# ----------------------------
# 3) Backtest per ticker (one trade at a time per ticker)
# ----------------------------
def backtest_one_ticker(
    g,
    buy_thresh=BUY_THRESH,
    stop_loss=STOP_LOSS,
    trail_start=TRAIL_START,
    trail_gap=TRAIL_GAP,
    h_max=H_MAX,
    fee=FEE_PER_TRADE,
):
    """
    g: single ticker df sorted by Date, must contain:
      Date, Ticker, Close, pred_max
    Enforces:
      - only 1 open position at a time (no repeated buys until exit)
    Exit rules:
      - hard stop
      - trailing stop after trail_start reached
      - "soft" horizon end: at >= h_max days, if trade is losing -> exit; if profitable -> keep holding
    """
    g = g.sort_values("Date").reset_index(drop=True)

    trades = []

    in_trade = False
    entry_date = None
    entry_price = None
    entry_pred = None
    target_price = None
    stop_price = None

    peak_price = None

    for i in range(len(g)):
        date = g.loc[i, "Date"]
        close = float(g.loc[i, "Close"])
        pred  = g.loc[i, "pred_max"]

        # skip rows without prediction
        if np.isnan(pred):
            continue

        if not in_trade:
            # ENTRY (only when flat)
            if float(pred) >= buy_thresh:
                in_trade = True
                entry_date = date
                entry_price = close
                entry_pred = float(pred)

                target_price = entry_price * (1 + entry_pred)
                stop_price   = entry_price * (1 + stop_loss)

                peak_price = entry_price

            continue

        # If IN TRADE:
        # update peak
        if close > peak_price:
            peak_price = close

        days_held = (date - entry_date).days
        gross_ret = close / entry_price - 1.0
        peak_ret  = peak_price / entry_price - 1.0

        # Trailing stop price (only active after peak_ret >= trail_start)
        trailing_price = None
        if peak_ret >= trail_start:
            trailing_price = peak_price * (1 - trail_gap)

        # Rule-exit price: active stop level (max of hard stop and trailing stop if active)
        if trailing_price is None:
            rule_exit_price = stop_price
        else:
            rule_exit_price = max(stop_price, trailing_price)

        exit_reason = None

        # 1) Hard stop / rule stop triggered
        if close <= rule_exit_price:
            exit_reason = "Stop/Trail"

        # 2) Soft horizon end behavior:
        #    - once we are beyond h_max days:
        #      * if losing -> exit
        #      * if profitable -> keep holding (no forced exit)
        elif days_held >= h_max and gross_ret <= 0:
            exit_reason = "SoftMaxHold(Losing)"

        if exit_reason is not None:
            # fee applied once on exit (simple). If you want entry+exit, use fee*2.
            net_ret = gross_ret - fee

            trades.append({
                "Ticker": g.loc[i, "Ticker"],
                "EntryDate": entry_date,
                "EntryPrice": entry_price,
                "ExitDate": date,
                "ExitPrice": close,
                "DaysHeld": days_held,

                "PredAtEntry": entry_pred,
                "TargetPrice": target_price,

                "StopPrice": stop_price,
                "PeakPrice": peak_price,
                "RuleExitPrice": rule_exit_price,

                "GrossRet": gross_ret,
                "NetRet": net_ret,
                "ExitReason": exit_reason,
            })

            # reset (flat again)
            in_trade = False
            entry_date = entry_price = entry_pred = None
            target_price = stop_price = None
            peak_price = None

    return pd.DataFrame(trades)


# ----------------------------
# 4) Run portfolio backtest (allows multiple tickers simultaneously)
#    We compute:
#      - trades_df: all trades
#      - by_ticker summary
#      - portfolio summary (equal-weight per ticker compounding)
# ----------------------------
def run_backtest(df_pred):
    tickers = sorted(df_pred["Ticker"].unique().tolist())

    all_trades = []
    per_ticker_final = []

    for t in tickers:
        g = df_pred[df_pred["Ticker"] == t].copy()
        if g.empty:
            continue

        trades_t = backtest_one_ticker(g)
        if not trades_t.empty:
            all_trades.append(trades_t)

            # Per-ticker compounding of NetRet
            eq = 1.0
            for r in trades_t["NetRet"].values:
                eq *= (1 + float(r))
            per_ticker_final.append({"Ticker": t, "FinalEquity": eq, "Trades": len(trades_t)})

        else:
            per_ticker_final.append({"Ticker": t, "FinalEquity": 1.0, "Trades": 0})

    trades_df = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()

    by_ticker = pd.DataFrame(per_ticker_final).sort_values("FinalEquity", ascending=False).reset_index(drop=True)

    # Portfolio equity (equal-weight across tickers)
    portfolio_equity = by_ticker["FinalEquity"].mean() if not by_ticker.empty else 1.0

    winning_trades = trades_df[trades_df["NetRet"] > 0]
    losing_trades = trades_df[trades_df["NetRet"] < 0]

    summary = {
        "Tickers": int(len(by_ticker)),
        "TotalTrades": int(trades_df.shape[0]) if not trades_df.empty else 0,
        "AvgTradesPerTicker": float(by_ticker["Trades"].mean()) if not by_ticker.empty else 0.0,
        "PortfolioEquity(MeanTickerEquity)": float(portfolio_equity),
        "PortfolioTotalReturn": float(portfolio_equity - 1.0),
        "WinRate": float((trades_df["NetRet"] > 0).mean()) if not trades_df.empty else np.nan,
        "AvgNetRetPerTrade": float(trades_df["NetRet"].mean()) if not trades_df.empty else np.nan,
        "MedianNetRetPerTrade": float(trades_df["NetRet"].median()) if not trades_df.empty else np.nan,
        "AvgNerRerPerWinTrade": float(winning_trades["NetRet"].mean()) if not winning_trades.empty else np.nan,
        "AvgNerRerPerLossTrade": float(losing_trades["NetRet"].mean()) if not losing_trades.empty else np.nan
    }

    return summary, trades_df, by_ticker


# ----------------------------
# 5) Phase 5 sanity check + driver
# ----------------------------
def phase5_sanity_check():
    print("🚀 Phase 5 sanity check: daily swing backtest (Option A)\n")

    # Load model/meta/features
    model, meta, feature_cols = load_model_and_meta()
    print(f"✅ Loaded model: {MODEL_PATH}")
    print(f"✅ Loaded meta : {META_PATH}")
    print(f"   #feature_cols: {len(feature_cols)}")

    df = load_daily_features()
    print(f"\n✅ Loaded features: {FEATURE_PATH}")
    print(f"   Rows: {len(df):,} | Cols: {len(df.columns)}")
    print(f"   Date range: {df['Date'].min()} → {df['Date'].max()}")
    print(f"   Tickers: {df['Ticker'].nunique()}")

    # Add predictions
    df_pred = add_predictions(df, model, feature_cols)
    ok_preds = df_pred["pred_max"].notna().mean()
    print(f"\n✅ Predictions added: pred_max")
    print(f"   % rows with pred_max: {ok_preds:.2%}")
    print(f"   pred_max stats (non-NaN):")
    print(df_pred["pred_max"].dropna().describe(percentiles=[0.1,0.5,0.9]).to_string())

    # Backtest
    summary, trades_df, by_ticker = run_backtest(df_pred)

    print("\n📌 Backtest summary:")
    for k, v in summary.items():
        if isinstance(v, float):
            print(f"   {k}: {v:.4f}")
        else:
            print(f"   {k}: {v}")

    # Ensure no overlapping trades per ticker (quick validation)
    if not trades_df.empty:
        overlaps = 0
        for t, g in trades_df.groupby("Ticker"):
            g = g.sort_values("EntryDate")
            # overlapping if next EntryDate <= prev ExitDate
            prev_exit = None
            for _, r in g.iterrows():
                if prev_exit is not None and r["EntryDate"] <= prev_exit:
                    overlaps += 1
                prev_exit = r["ExitDate"]
        print(f"\n🧪 Overlap check (should be 0): {overlaps}")

        print("\n🧾 Sample trades (first 10):")
        show_cols = ["Ticker","EntryDate","EntryPrice","PredAtEntry","TargetPrice",
                     "ExitDate","ExitPrice","RuleExitPrice","NetRet","ExitReason","DaysHeld", "StopPrice"]
        print(trades_df[show_cols].head(10).to_string(index=False))

        print("\n🏁 Top 10 tickers by final equity:")
        print(by_ticker.head(10).to_string(index=False))
    else:
        print("\n⚠️ No trades triggered. Consider lowering BUY_THRESH or checking pred_max distribution.")

    # Save outputs
    os.makedirs(os.path.dirname(OUT_TRADES_PATH), exist_ok=True)
    if not trades_df.empty:
        trades_df.to_csv(OUT_TRADES_PATH, index=False)
    by_ticker.to_csv(OUT_EQUITY_PATH, index=False)

    print(f"\n💾 Saved:")
    print(f"   Trades  → {OUT_TRADES_PATH}")
    print(f"   Equity  → {OUT_EQUITY_PATH}")

    return summary, trades_df, by_ticker


# ----------------------------
# RUN PHASE 5
# ----------------------------
summary, trades_df_swing, by_ticker_swing = phase5_sanity_check()
```

```python
#Single position Backtest
import os, json, pickle
import numpy as np
import pandas as pd

# ----------------------------
# Paths (match your project)
# ----------------------------
FEATURE_PATH = "/content/drive/MyDrive/SMA_Swing_Project/daily_features_live.parquet"
MODEL_PATH   = "/content/drive/MyDrive/SMA_Swing_Project/xgb_swing_model_live.pkl"
META_PATH    = "/content/drive/MyDrive/SMA_Swing_Project/xgb_swing_model_meta_live.json"

OUT_TRADES_PATH = "/content/drive/MyDrive/SMA_Swing_Project/backtest_trades_phase5_1.csv"
OUT_EQUITY_PATH = "/content/drive/MyDrive/SMA_Swing_Project/backtest_equity_phase5_1.csv"

# ----------------------------
# Strategy parameters (tune)
# ----------------------------
BUY_THRESH     = 0.325    # NOTE: 0.175 = 17.5%. If you meant 1.5% use 0.015
STOP_LOSS      = -0.1   # hard stop at -7.5% from entry

TRAIL_START    = 0.1     # start trailing once trade has been +10% at any time
TRAIL_GAP      = 0.025     # allow 3% drawdown from peak once trailing is active

H_MIN = 5
H_MAX = 10

FEE_PER_TRADE  = 0.000    # subtract on exit (simple)

# ----------------------------
# 0) Load model + meta
# ----------------------------
def load_model_and_meta(model_path=MODEL_PATH, meta_path=META_PATH):
    with open(model_path, "rb") as f:
        model = pickle.load(f)
    with open(meta_path, "r") as f:
        meta = json.load(f)

    feature_cols = meta.get("feature_cols", None)
    if not feature_cols or not isinstance(feature_cols, list):
        raise ValueError("META missing feature_cols (list).")

    return model, meta, feature_cols


# ----------------------------
# 1) Load features
# ----------------------------
def load_daily_features(feature_path=FEATURE_PATH):
    df = pd.read_parquet(feature_path)
    if "Date" not in df.columns:
        raise ValueError("Feature dataset must include 'Date'.")
    df["Date"] = pd.to_datetime(df["Date"])
    df = df.sort_values(["Ticker", "Date"]).reset_index(drop=True)
    return df


# ----------------------------
# 2) Add predictions (pred_max)
# ----------------------------
def add_predictions(df, model, feature_cols):
    missing = [c for c in feature_cols if c not in df.columns]
    if missing:
        raise KeyError(f"Feature dataset missing required feature_cols: {missing}")

    X = df[feature_cols].apply(pd.to_numeric, errors="coerce")
    ok = ~X.isna().any(axis=1)

    df = df.copy()
    df["pred_max"] = np.nan

    if ok.sum() == 0:
        raise ValueError("No rows have complete numeric features for prediction.")

    preds = model.predict(X.loc[ok].values)
    df.loc[ok, "pred_max"] = preds.astype(float)
    return df


# ----------------------------
# 3) Backtest: SINGLE POSITION TOTAL (portfolio holds 0 or 1 ticker)
# ----------------------------
def run_backtest_single_position(
    df_pred,
    buy_thresh=BUY_THRESH,
    stop_loss=STOP_LOSS,
    trail_start=TRAIL_START,
    trail_gap=TRAIL_GAP,
    h_max=H_MAX,
    fee=FEE_PER_TRADE,
):
    """
    ONE position total across the entire universe.

    Entry:
      - when flat, each day pick the ticker with highest pred_max among those >= buy_thresh

    Exit rules (same as your per-ticker logic):
      - Stop/Trail: close <= rule_exit_price
      - SoftMaxHold(Losing): if days_held >= h_max and gross_ret <= 0 -> exit
      - If profitable at h_max, keep holding (no forced exit)
    """

    df = df_pred.copy()
    df["Date"] = pd.to_datetime(df["Date"])
    df = df.sort_values(["Date", "Ticker"]).reset_index(drop=True)

    required = {"Date", "Ticker", "Close", "pred_max"}
    missing_req = required - set(df.columns)
    if missing_req:
        raise ValueError(f"df_pred missing required columns: {sorted(missing_req)}")

    all_dates = sorted(df["Date"].unique())

    # portfolio state
    in_trade = False
    pos_ticker = None
    entry_date = None
    entry_price = None
    entry_pred = None
    target_price = None
    stop_price = None
    peak_price = None

    equity = 1.0
    trades = []
    equity_rows = []

    for d in all_dates:
        # Save equity snapshot each date
        equity_rows.append({
            "Date": d,
            "Equity": equity,
            "Holding": pos_ticker if in_trade else ""
        })

        # ----------------
        # ENTRY (only if flat)
        # ----------------
        if not in_trade:
            day = df[df["Date"] == d][["Ticker", "Close", "pred_max"]].copy()
            day = day.dropna(subset=["Close", "pred_max"])

            # candidates that pass threshold
            cand = day[day["pred_max"].astype(float) >= buy_thresh]
            if cand.empty:
                continue

            # pick highest possibility = highest pred_max
            best = cand.sort_values("pred_max", ascending=False).iloc[0]

            pos_ticker   = str(best["Ticker"])
            entry_date   = d
            entry_price  = float(best["Close"])
            entry_pred   = float(best["pred_max"])

            target_price = entry_price * (1 + entry_pred)
            stop_price   = entry_price * (1 + stop_loss)
            peak_price   = entry_price

            in_trade = True
            continue

        # ----------------
        # IN TRADE: manage only the held ticker
        # ----------------
        row = df[(df["Date"] == d) & (df["Ticker"].astype(str) == pos_ticker)]
        if row.empty:
            continue

        close = float(row["Close"].iloc[0])

        if close > peak_price:
            peak_price = close

        days_held = (d - entry_date).days
        gross_ret = close / entry_price - 1.0
        peak_ret  = peak_price / entry_price - 1.0

        trailing_price = None
        if peak_ret >= trail_start:
            trailing_price = peak_price * (1 - trail_gap)

        rule_exit_price = stop_price if trailing_price is None else max(stop_price, trailing_price)

        exit_reason = None
        if close <= rule_exit_price:
            exit_reason = "Stop/Trail"
        elif days_held >= h_max and gross_ret <= 0:
            exit_reason = "SoftMaxHold(Losing)"

        if exit_reason is not None:
            net_ret = gross_ret - fee
            equity *= (1 + net_ret)

            trades.append({
                "Ticker": pos_ticker,
                "EntryDate": entry_date,
                "EntryPrice": entry_price,
                "ExitDate": d,
                "ExitPrice": close,
                "DaysHeld": days_held,

                "PredAtEntry": entry_pred,
                "TargetPrice": target_price,

                "StopPrice": stop_price,
                "PeakPrice": peak_price,
                "RuleExitPrice": rule_exit_price,

                "GrossRet": gross_ret,
                "NetRet": net_ret,
                "ExitReason": exit_reason,
            })

            # reset to flat
            in_trade = False
            pos_ticker = None
            entry_date = entry_price = entry_pred = None
            target_price = stop_price = None
            peak_price = None

    trades_df = pd.DataFrame(trades)
    equity_df = pd.DataFrame(equity_rows)

    # by-ticker (from actual executed trades)
    if not trades_df.empty:
        by_ticker = (trades_df.groupby("Ticker")
                     .agg(
                         Trades=("NetRet", "size"),
                         WinRate=("NetRet", lambda x: float((x > 0).mean())),
                         AvgNetRet=("NetRet", "mean"),
                         MedianNetRet=("NetRet", "median"),
                         TotalNetRet=("NetRet", lambda x: float(np.prod(1 + x.values) - 1)),
                     )
                     .reset_index()
                     .sort_values("TotalNetRet", ascending=False)
                     .reset_index(drop=True))
    else:
        by_ticker = pd.DataFrame(columns=["Ticker","Trades","WinRate","AvgNetRet","MedianNetRet","TotalNetRet"])

    summary = {
        "Mode": "SinglePositionPortfolio",
        "TotalTrades": int(len(trades_df)),
        "FinalEquity": float(equity),
        "TotalReturn": float(equity - 1.0),
        "WinRate": float((trades_df["NetRet"] > 0).mean()) if not trades_df.empty else np.nan,
        "AvgNetRetPerTrade": float(trades_df["NetRet"].mean()) if not trades_df.empty else np.nan,
        "MedianNetRetPerTrade": float(trades_df["NetRet"].median()) if not trades_df.empty else np.nan,
    }

    return summary, trades_df, equity_df, by_ticker


# ----------------------------
# 4) Phase 5 sanity check + driver
# ----------------------------
def phase5_sanity_check():
    print("🚀 Phase 5 sanity check: daily swing backtest (SINGLE POSITION)\n")

    model, meta, feature_cols = load_model_and_meta()
    print(f"✅ Loaded model: {MODEL_PATH}")
    print(f"✅ Loaded meta : {META_PATH}")
    print(f"   #feature_cols: {len(feature_cols)}")

    df = load_daily_features()
    print(f"\n✅ Loaded features: {FEATURE_PATH}")
    print(f"   Rows: {len(df):,} | Cols: {len(df.columns)}")
    print(f"   Date range: {df['Date'].min()} → {df['Date'].max()}")
    print(f"   Tickers: {df['Ticker'].nunique()}")

    df_pred = add_predictions(df, model, feature_cols)
    ok_preds = df_pred["pred_max"].notna().mean()
    print(f"\n✅ Predictions added: pred_max")
    print(f"   % rows with pred_max: {ok_preds:.2%}")
    print("   pred_max stats (non-NaN):")
    print(df_pred["pred_max"].dropna().describe(percentiles=[0.1,0.5,0.9]).to_string())

    summary, trades_df, equity_df, by_ticker = run_backtest_single_position(df_pred)

    print("\n📌 Backtest summary:")
    for k, v in summary.items():
        if isinstance(v, float):
            print(f"   {k}: {v:.4f}")
        else:
            print(f"   {k}: {v}")

    # Sanity: there should never be overlapping trades (by definition)
    if not trades_df.empty:
        trades_sorted = trades_df.sort_values("EntryDate").reset_index(drop=True)
        overlaps = 0
        prev_exit = None
        for _, r in trades_sorted.iterrows():
            if prev_exit is not None and r["EntryDate"] <= prev_exit:
                overlaps += 1
            prev_exit = r["ExitDate"]
        print(f"\n🧪 Overlap check (should be 0): {overlaps}")

        print("\n🧾 Sample trades (first 10):")
        show_cols = ["Ticker","EntryDate","EntryPrice","PredAtEntry","TargetPrice",
                     "ExitDate","ExitPrice","RuleExitPrice","NetRet","ExitReason","DaysHeld"]
        print(trades_df[show_cols].head(10).to_string(index=False))

        print("\n🏁 Most traded tickers / best total net return tickers:")
        print(by_ticker.head(10).to_string(index=False))
    else:
        print("\n⚠️ No trades triggered. Consider lowering BUY_THRESH or checking pred_max distribution.")

    # Save outputs
    os.makedirs(os.path.dirname(OUT_TRADES_PATH), exist_ok=True)
    if not trades_df.empty:
        trades_df.to_csv(OUT_TRADES_PATH, index=False)
    equity_df.to_csv(OUT_EQUITY_PATH, index=False)

    print(f"\n💾 Saved:")
    print(f"   Trades  → {OUT_TRADES_PATH}")
    print(f"   Equity  → {OUT_EQUITY_PATH}")

    return summary, trades_df, equity_df, by_ticker


# ----------------------------
# RUN PHASE 5
# ----------------------------
summary, trades_df_swing, equity_df_swing, by_ticker_swing = phase5_sanity_check()
```

```python
 # ==========================================
# PHASE 6: LIVE DAILY SWING SIGNAL ENGINE
#        (FINNHUB, MULTI-POSITION)
#        ✅ MANUAL BUY + MANUAL CLOSE (NO WATCHLIST)
# ==========================================

import os, json, pickle, time, datetime
import numpy as np
import pandas as pd
import pytz

# ----------------------------
# Paths
# ----------------------------
MODEL_PATH   = "/content/drive/MyDrive/SMA_Swing_Project/xgb_swing_model_live.pkl"
META_PATH    = "/content/drive/MyDrive/SMA_Swing_Project/xgb_swing_model_meta_live.json"

LIVE_STATE_PATH   = "/content/drive/MyDrive/SMA_Swing_Project/live_state_multi_free.json"
LIVE_SIGNALS_CSV  = "/content/drive/MyDrive/SMA_Swing_Project/live_signals_log_free.csv"
LIVE_TRADES_CSV   = "/content/drive/MyDrive/SMA_Swing_Project/live_trades_log_free.csv"

# ----------------------------
# Strategy parameters
# ----------------------------
BUY_THRESH    = 0.20
STOP_LOSS     = -0.1
TRAIL_START   = 0.10
TRAIL_GAP     = 0.025
H_MAX         = 10
FEE_PER_TRADE = 0.000

# ----------------------------
# Live loop params
# ----------------------------
CHECK_EVERY_SEC = 1 * 60
ITERATIONS      = 24 * 7
LOOKBACK_DAYS   = 365 * 1

TIMEZONE = "US/Eastern"
tz_et = pytz.timezone(TIMEZONE)

# ----------------------------
# Manual control ✅
# ----------------------------
MANUAL_SELECT = True   # choose what to BUY each cycle
SHOW_TOP_N    = 10     # show top N candidates

# =========================================================
# 0) REQUIREMENTS
# =========================================================
try:
    finnhub_client
except NameError:
    raise NameError("finnhub_client not found. Make sure Phase 1 created finnhub_client.")

try:
    SYMBOLS
except NameError:
    raise NameError("SYMBOLS not found. Make sure Phase 1 defined SYMBOLS.")

# =========================================================
# 1) Load model + meta
# =========================================================
with open(MODEL_PATH, "rb") as f:
    model = pickle.load(f)

with open(META_PATH, "r") as f:
    meta = json.load(f)

feature_cols = meta.get("feature_cols", None)
if not feature_cols:
    raise ValueError("META missing feature_cols.")

print(f"✅ Loaded model: {MODEL_PATH}")
print(f"✅ Loaded meta : {META_PATH}")
print(f"   #feature_cols: {len(feature_cols)}")

# =========================================================
# 2) State persistence
# =========================================================
def load_live_state(path=LIVE_STATE_PATH):
    if os.path.exists(path):
        with open(path, "r") as f:
            return json.load(f)

    return {
        "open_positions": {},
        "last_run_et": None,
    }

def save_live_state(state, path=LIVE_STATE_PATH):
    os.makedirs(os.path.dirname(path), exist_ok=True)
    with open(path, "w") as f:
        json.dump(state, f, indent=2, default=str)

# =========================================================
# 3) Logging helpers
# =========================================================
def append_csv_row(path, row_dict):
    os.makedirs(os.path.dirname(path), exist_ok=True)
    exists = os.path.exists(path)
    df_row = pd.DataFrame([row_dict])
    df_row.to_csv(path, mode="a", header=not exists, index=False)

def log_signal_snapshot(snapshot):
    append_csv_row(LIVE_SIGNALS_CSV, snapshot)

def log_trade_action(trade_row):
    append_csv_row(LIVE_TRADES_CSV, trade_row)

# =========================================================
# 4) Finnhub daily fetch
# =========================================================
def fetch_finnhub_daily(symbol, lookback_days=LOOKBACK_DAYS):
    now = int(time.time())
    past = now - int(lookback_days * 24 * 60 * 60)

    js = finnhub_client.stock_candles(symbol, "D", past, now)
    status = js.get("s", None)
    if status != "ok":
        return None, status

    df = pd.DataFrame({
        "Date":   pd.to_datetime(js["t"], unit="s", utc=True),
        "Open":   js["o"],
        "High":   js["h"],
        "Low":    js["l"],
        "Close":  js["c"],
        "Volume": js["v"],
    })
    if df.empty:
        return None, "empty"

    df["Date"] = df["Date"].dt.tz_convert(TIMEZONE).dt.normalize()
    df["Ticker"] = symbol
    df = df.sort_values("Date").reset_index(drop=True)

    for c in ["Open","High","Low","Close","Volume"]:
        df[c] = pd.to_numeric(df[c], errors="coerce")

    return df, "ok"

# =========================================================
# 5) Feature builder (Phase 3 exact + legacy aliases)
# =========================================================
def add_live_features_phase3_exact(
    daily_df,
    sma_windows_short=(5, 10, 20),
    sma_windows_long=(50, 100),
    ema_windows=(5, 10, 20),
    vol_windows=(10, 20, 50),
    rsi_period=14,
    bb_window=20,
    bb_k=2.0,
):
    df = daily_df.sort_values(["Ticker", "Date"]).copy()
    out_frames = []

    for ticker, g in df.groupby("Ticker"):
        g = g.sort_values("Date").reset_index(drop=True).copy()

        g["ret_1d"] = g["Close"].pct_change()
        g["log_ret_1d"] = np.log(g["Close"]).diff()

        # legacy aliases (model meta expects)
        g["Adj Close"] = g["Close"]
        g["Ret_1"] = g["ret_1d"]
        g["LogRet_1"] = g["log_ret_1d"]
        g["DayOfWeek"] = g["Date"].dt.dayofweek
        g["DayIndex"] = np.arange(len(g))

        for w in sma_windows_short + sma_windows_long:
            g[f"sma_{w}"] = g["Close"].rolling(w).mean()

        for w in ema_windows:
            g[f"ema_{w}"] = g["Close"].ewm(span=w, adjust=False).mean()

        for w in vol_windows:
            g[f"vol_{w}"] = g["ret_1d"].rolling(w).std()

        g[f"bb_mid_{bb_window}"] = g["Close"].rolling(bb_window).mean()
        g[f"bb_std_{bb_window}"] = g["Close"].rolling(bb_window).std()
        g[f"bb_up_{bb_window}"]  = g[f"bb_mid_{bb_window}"] + bb_k * g[f"bb_std_{bb_window}"]
        g[f"bb_dn_{bb_window}"]  = g[f"bb_mid_{bb_window}"] - bb_k * g[f"bb_std_{bb_window}"]

        delta = g["Close"].diff()
        gain = delta.clip(lower=0).rolling(rsi_period).mean()
        loss = (-delta.clip(upper=0)).rolling(rsi_period).mean()
        rs = gain / loss
        g[f"rsi_{rsi_period}"] = 100 - (100 / (1 + rs))

        out_frames.append(g)

    return pd.concat(out_frames, ignore_index=True)

# =========================================================
# 6) Predict last valid row
# =========================================================
def predict_for_symbol(symbol):
    df_raw, status = fetch_finnhub_daily(symbol, lookback_days=LOOKBACK_DAYS)
    if df_raw is None:
        return None, f"{symbol}: {status}"

    g = add_live_features_phase3_exact(df_raw)

    missing = [c for c in feature_cols if c not in g.columns]
    if missing:
        return None, f"{symbol}: missing features {missing[:10]}"

    X = g[feature_cols].apply(pd.to_numeric, errors="coerce")
    ok = ~X.isna().any(axis=1)
    if ok.sum() == 0:
        return None, f"{symbol}: no valid feature rows (NaNs)"

    last = g.loc[ok].iloc[-1].copy()
    X_last = X.loc[ok].iloc[-1].astype("float32").values.reshape(1, -1)
    pred = float(model.predict(X_last)[0])

    last["pred_max"] = pred
    return last, "ok"

# =========================================================
# 7) Exit rules
# =========================================================
def compute_rule_exit_price(entry_price, stop_price, peak_price):
    peak_ret = peak_price / entry_price - 1.0
    trailing_price = None
    if peak_ret >= TRAIL_START:
        trailing_price = peak_price * (1 - TRAIL_GAP)
    rule_exit = stop_price if trailing_price is None else max(stop_price, trailing_price)
    return rule_exit, trailing_price

# =========================================================
# ✅ MANUAL COMMANDS (BUY / CLOSE)
# =========================================================
def parse_index_list(txt):
    nums = []
    for part in txt.replace(" ", "").split(","):
        if part.isdigit():
            nums.append(int(part))
    return nums

def manual_command(open_pos_keys, candidates_sorted):
    """
    Commands:
      - buy 1,3
      - close NVDA
      - close 1,2   (by open position list index)
      - close all
      - skip
    """
    print("\n🎮 COMMANDS:")
    print("  buy 1,3         -> open paper positions from candidates")
    print("  close NVDA      -> force close that ticker")
    print("  close 1,2       -> close by position list index")
    print("  close all       -> close everything")
    print("  skip            -> do nothing\n")

    cmd = input("👉 Enter command: ").strip()
    if not cmd:
        return {"buy": [], "close": []}

    cmd_l = cmd.lower().strip()

    # Skip
    if cmd_l in ["skip", "none"]:
        return {"buy": [], "close": []}

    # Close all
    if cmd_l == "close all":
        return {"buy": [], "close": ["__ALL__"]}

    # Close by ticker name
    if cmd_l.startswith("close "):
        arg = cmd.split(" ", 1)[1].strip()
        # close by indices
        if any(ch.isdigit() for ch in arg) and "," in arg or arg.isdigit():
            idxs = parse_index_list(arg)
            close_list = []
            for i in idxs:
                if 1 <= i <= len(open_pos_keys):
                    close_list.append(open_pos_keys[i-1])
            return {"buy": [], "close": close_list}
        else:
            # close by symbol
            sym = arg.upper()
            return {"buy": [], "close": [sym]}

    # Buy
    if cmd_l.startswith("buy "):
        arg = cmd.split(" ", 1)[1].strip()
        idxs = parse_index_list(arg)
        buy_list = []
        for i in idxs:
            if 1 <= i <= len(candidates_sorted):
                buy_list.append(candidates_sorted[i-1])
        return {"buy": buy_list, "close": []}

    print("⚠️ Unknown command -> skip")
    return {"buy": [], "close": []}

# =========================================================
# ✅ Force close function
# =========================================================
def force_close_position(open_pos, sym, now_et):
    last, status = predict_for_symbol(sym)
    if last is None:
        print(f"⚠️ Cannot close {sym}: data error {status}")
        return False

    cur_date = pd.Timestamp(last["Date"]).date()
    close_price = float(last["Close"])

    entry_date = open_pos[sym]["entry_date"]
    entry_price = float(open_pos[sym]["entry_price"])
    entry_pred  = float(open_pos[sym]["entry_pred"])
    target_price = float(open_pos[sym]["target_price"])
    stop_price  = float(open_pos[sym]["stop_price"])
    peak_price  = float(open_pos[sym]["peak_price"])

    days_held = (pd.to_datetime(cur_date) - pd.to_datetime(entry_date)).days
    gross_ret = close_price / entry_price - 1.0
    net_ret   = gross_ret - FEE_PER_TRADE

    log_trade_action({
        "TimestampET": now_et,
        "Action": "SELL",
        "Ticker": sym,
        "EntryDate": entry_date,
        "EntryPrice": entry_price,
        "ExitDate": str(cur_date),
        "ExitPrice": close_price,
        "DaysHeld": days_held,
        "PredAtEntry": entry_pred,
        "TargetPrice": target_price,
        "StopPrice": stop_price,
        "PeakPrice": peak_price,
        "RuleExitPrice": np.nan,
        "GrossRet": gross_ret,
        "NetRet": net_ret,
        "ExitReason": "ManualClose",
    })

    del open_pos[sym]
    print(f"✅ MANUAL CLOSE: {sym} sold @ {close_price:.2f} logged → {LIVE_TRADES_CSV}")
    return True

# =========================================================
# 8) Live cycle
# =========================================================
def live_decision_cycle():
    state = load_live_state()
    open_pos = state.get("open_positions", {})

    now_et = datetime.datetime.now(tz_et).strftime("%Y-%m-%d %H:%M:%S")

    # -----------------------------
    # A) POSITION CHECK
    # -----------------------------
    if open_pos:
        print(f"\n[{now_et}] 📦 OPEN POSITIONS ({len(open_pos)} total)")
        open_keys = sorted(list(open_pos.keys()))
        for idx, sym in enumerate(open_keys, start=1):
            last, status = predict_for_symbol(sym)
            if last is None:
                print(f"  [{idx}] {sym}: ⚠️ data error {status}")
                continue

            cur_date = pd.Timestamp(last["Date"]).date()
            close = float(last["Close"])

            entry_date = pd.to_datetime(open_pos[sym]["entry_date"]).date()
            entry_price = float(open_pos[sym]["entry_price"])
            peak_price  = float(open_pos[sym]["peak_price"])
            stop_price  = float(open_pos[sym]["stop_price"])

            if close > peak_price:
                peak_price = close
                open_pos[sym]["peak_price"] = peak_price

            gross_ret = close / entry_price - 1.0
            rule_exit, _ = compute_rule_exit_price(entry_price, stop_price, peak_price)

            print(f"  [{idx}] {sym}: Entry={entry_price:.2f} | Now={close:.2f} | P/L={gross_ret:.2%} | RuleExit={rule_exit:.2f}")

    else:
        open_keys = []

    # -----------------------------
    # B) SCAN BUY CANDIDATES
    # -----------------------------
    candidates = []
    errors = 0
    error_examples = []

    for sym in SYMBOLS:
        if sym in open_pos:
            continue

        last, status = predict_for_symbol(sym)
        if last is None:
            errors += 1
            if len(error_examples) < 5:
                error_examples.append(status)
            continue

        pred = float(last["pred_max"])
        if pred >= BUY_THRESH:
            close = float(last["Close"])
            candidates.append({
                "Ticker": sym,
                "Date": str(pd.Timestamp(last["Date"]).date()),
                "EntryPrice": close,
                "PredMax": pred,
                "TargetPrice": close * (1 + pred),
                "StopPrice": close * (1 + STOP_LOSS),
            })

    if candidates:
        candidates_sorted = sorted(candidates, key=lambda x: x["PredMax"], reverse=True)
        top_show = candidates_sorted[:int(SHOW_TOP_N)]

        print(f"\n[{now_et}] 🟢 BUY CANDIDATES (PredMax >= {BUY_THRESH:.2%}) | count={len(candidates_sorted)} | errors={errors}")
        for idx, c in enumerate(top_show, start=1):
            print(f"  [{idx}] {c['Date']} {c['Ticker']}: Entry={c['EntryPrice']:.2f} | PredMax={c['PredMax']:.2%} | "
                  f"Target={c['TargetPrice']:.2f} | Stop={c['StopPrice']:.2f}")

        # log candidates
        for c in top_show:
            log_signal_snapshot({
                "TimestampET": now_et,
                "Mode": "SCAN",
                "Action": "BUY_CANDIDATE",
                "Ticker": c["Ticker"],
                "Date": c["Date"],
                "EntryDate": c["Date"],
                "EntryPrice": c["EntryPrice"],
                "CurrentClose": c["EntryPrice"],
                "DaysHeld": 0,
                "PredAtEntry": c["PredMax"],
                "PredNow": c["PredMax"],
                "TargetPrice": c["TargetPrice"],
                "StopPrice": c["StopPrice"],
                "PeakPrice": c["EntryPrice"],
                "TrailingStop": np.nan,
                "RuleExitPrice": c["StopPrice"],
                "ExitReason": "",
            })

    else:
        candidates_sorted = []
        print(f"\n[{now_et}] 🧊 No BUY candidates (PredMax >= {BUY_THRESH:.2%}) | errors={errors}")
        if error_examples:
            print("   ⚠️ Error examples:")
            for ex in error_examples:
                print("    -", ex)

    # -----------------------------
    # ✅ MANUAL ACTION STEP (BUY / CLOSE)
    # -----------------------------
    if MANUAL_SELECT:
        cmd = manual_command(open_keys, candidates_sorted[:int(SHOW_TOP_N)])

        # CLOSE
        close_list = cmd["close"]
        if close_list:
            if close_list == ["__ALL__"]:
                for sym in list(open_pos.keys()):
                    force_close_position(open_pos, sym, now_et)
            else:
                for sym in close_list:
                    if sym in open_pos:
                        force_close_position(open_pos, sym, now_et)
                    else:
                        print(f"⚠️ {sym} not in open_positions")

        # BUY
        buy_list = cmd["buy"]
        if buy_list:
            for c in buy_list:
                sym = c["Ticker"]
                if sym in open_pos:
                    continue

                open_pos[sym] = {
                    "entry_date": c["Date"],
                    "entry_price": float(c["EntryPrice"]),
                    "entry_pred": float(c["PredMax"]),
                    "target_price": float(c["TargetPrice"]),
                    "stop_price": float(c["StopPrice"]),
                    "peak_price": float(c["EntryPrice"]),
                }

                log_trade_action({
                    "TimestampET": now_et,
                    "Action": "BUY",
                    "Ticker": sym,
                    "EntryDate": c["Date"],
                    "EntryPrice": float(c["EntryPrice"]),
                    "PredAtEntry": float(c["PredMax"]),
                    "TargetPrice": float(c["TargetPrice"]),
                    "StopPrice": float(c["StopPrice"]),
                    "PeakPrice": float(c["EntryPrice"]),
                    "RuleExitPrice": float(c["StopPrice"]),
                    "Reason": f"Manual BUY (PredMax >= {BUY_THRESH:.2%})",
                })

                print(f"✅ ENTERED (paper): {sym}")

    # Save state
    state["open_positions"] = open_pos
    state["last_run_et"] = now_et
    save_live_state(state)

# =========================================================
# 9) Live runner
# =========================================================
def run_live_swing(iterations=ITERATIONS, sleep_sec=CHECK_EVERY_SEC):
    print("\n🚀 Starting LIVE swing engine (daily candles, manual BUY + manual CLOSE)")
    print(f"   Check every: {sleep_sec/60:.1f} min | Iterations: {iterations}")
    print(f"   Logs:\n    - {LIVE_SIGNALS_CSV}\n    - {LIVE_TRADES_CSV}\n    - {LIVE_STATE_PATH}\n")

    for i in range(iterations):
        now_et = datetime.datetime.now(tz_et).strftime("%Y-%m-%d %H:%M:%S")
        try:
            live_decision_cycle()
        except Exception as e:
            print(f"[{now_et}] ⚠️ Unexpected error: {type(e).__name__}: {e}")

        now_et = datetime.datetime.now(tz_et).strftime("%Y-%m-%d %H:%M:%S")
        print(f"\n[{now_et}] ✅ Heartbeat {i+1}/{iterations} | Sleeping...\n")
        time.sleep(sleep_sec)

# ----------------------------
# RUN LIVE
# ----------------------------
run_live_swing(iterations=ITERATIONS, sleep_sec=CHECK_EVERY_SEC)

```
